### Questão 01
Determine uma função 𝑓(𝑛) para o código a seguir tal que 𝑇(𝑛)=𝜃(𝑓(𝑛)). Explique como a 
solução foi encontrada.

```c
int compute(int v[], int left, int right) { 
    int n = right - left + 1; 
    if (n <= 1) return 0; 
     
    int mid = (left + right) / 2; 
     
    int sum = 0; 
    sum += compute(v, left, mid); 
    sum += compute(v, mid + 1, right); 
    sum += compute(v, left + n/4, mid + n/4); 
    sum += compute(v, mid - n/4, right - n/4); 
     
    int temp[n]; 
    int tempSize = 0;  
    for (int i = left; i <= right; i++) { 
        int pos = 0; 
        while (pos < tempSize && temp[pos] < v[i]) { 
            pos++; 
        } 
         
        for (int j = tempSize; j > pos; j--) { 
            temp[j] = temp[j-1]; 
        } 
         
        temp[pos] = v[i]; 
        tempSize++; 
        sum++; 
    } 
     
    for (int i = 0; i < n; i++) { 
        v[left + i] = temp[i]; 
    } 
     
    return sum; 
}
```

A função `compute` define um caso base `n<=1` tal que $T(1)$ é $\Theta(1)$.

Em outros casos, a função `compute` realiza os seguintes procedimentos: 
- Operações aritiméticas (calculo de `n` e `mid`) e inicialização variàveis (`temp[n]`e `tempSize`), as quais tem custo constante;
- Quatro chamadas recusivas com tamanho $\frac{n}{2}$ cada;
- Uma iteração sobre os $n$ elementos da entrada onde, em um passo $k$, se realizam as seguitne operações:
    - Um loop `while` sobre um número $p$ de elementos, com $0 \le p \le k$;
    - Um loop `for` sobre os elementos entre $k$ e $p$;
    - Operações aritiméticas de tempo constante;
    - Totalizando um custo na ordem de $k$
- Uma iteração sobre os $n$ elementos da entrada, atualizando valores.

Logo, temos que o custo da função `compute` para uma entrada de tamaho $n$ será:
$$
T(n) = 4 T(\frac{n}{2}) + \sum_{k=0}^{n}{(k + 1)} + n + 1 = 4 T(\frac{n}{2}) + \frac{n(n+1)}{2} + n + n = 4 T(\frac{n}{2}) + \frac{n^2+n}{2} + 2n = 4 T(\frac{n}{2}) + \frac{n^2 + 5n}{2}
$$

Considerando a notação $T(n) = a T(\frac{n}{b}) + f(n)$ usada como referência no Teorema Mestre, temos que $a=4$, $b=2$ e $f(n) = \frac{n^2 + 5n}{2}$. Logo, como $\log_b{a} = \log_2{4} = 2$ e $f(n) = \Theta(n^2\cdot\log^0{n})$, temos que $f(n) = \Theta(n^{\log_b{a}}\cdot\log^k{n})$ e estamos no caso $2$ do Teorema Mestre, de forma que $T(n) = \Theta(n^{\log_b{a}}\cdot\log^{k+1}{n})=\Theta(n^2\cdot\log{n})$.

### Questão 02
Uma empresa de logística mantém produtos em um armazém circular organizados em ordem 
crescente. Cada produto é identificado por um código numérico. Quando o armazém sofre 
rotação para otimização de espaço, a sequência ordenada é "rotacionada" - alguns elementos 
do final vão para o início, mantendo a ordem relativa. Por exemplo, a sequência [10, 20, 30, 40, 
50] após rotação de 2 posições se torna [40, 50, 10, 20, 30]. Desenvolva um algoritmo eficiente 
para encontrar um produto específico neste armazém rotacionado, retornando a sua posição 
ou -1 caso não exista. O algoritmo deverá ter complexidade 𝑂(𝑙𝑜𝑔 𝑛). 

**Exemplos:**

| Entrada | Saída |
|---|---|
| `produtos = [40, 50, 60, 10, 20, 30]` e $k=10$ | 3 |
| `produtos = [4, 5, 6, 7, 0, 1, 2] ` e $k=0$ | 4 |
| `produtos = [30, 40, 50, 10, 20] ` e $k=35$ | -1 |

````python
def find_min_pos(v: list, low: int, high: int) -> int:
    """Retorna o índice do menor valor de um vetor crescente de ids únicos que sofreu uma rotação rotação

    Parameters
    ----------
    v : list
        entradas crescentes de ids únicos a menos de um offset
    low : int
        ínicio do intervalo de busca
    high : int
        fim do intervalo de busca

    Returns
    -------
    int
        índice do menor valor (tamanho do offset)
    """

    if (high <= low + 1):
        return low if (v[high] > v[low]) else high

    mid = low + (high - low) // 2 # mid >= low + 1  
    
    if (v[mid] <= v[low]): # o shift está entre low e mid
        return find_min_pos(v, low, mid)

    if (v[mid] >= v[high]): # o shift está entre mid e high
        return find_min_pos(v, mid+1, high)

    # caso v[low] < v[mid] < v[high], v[low:high+1] está ordenado, pois as estradas são ids únicos crescentes
    return low

def circular_binary_search(v, offset, logic_low, logic_high, val):
    n = len(v)
    if (logic_low == logic_high):
        return (offset+logic_low)%n if (v[(offset+logic_low)%n] == val) else -1

    logic_mid = logic_low + (logic_high-logic_low)//2
    real_mid = (offset + logic_mid) % n
    
    if (v[real_mid] == val):
        return real_mid
    if (v[real_mid] > val):
        return circular_binary_search(v, offset, logic_low, logic_mid, val)
    else: # (v[real_mid] < val)
        return circular_binary_search(v, offset, logic_mid+1, logic_high, val)

def questao_02(v, k):
    offset = find_min_pos(v, 0, len(v)-1)
    return circular_binary_search(v, offset, 0, len(v)-1, k)

print(questao_02([40, 50, 60, 10, 20, 30], 10))
print(questao_02([4, 5, 6, 7, 0, 1, 2], 0))
print(questao_02([30, 40, 50, 10, 20], 35))
````

### Questão 03
Um serviço de streaming precisa manter atualizado em tempo real um ranking com as 𝑘 
músicas mais tocadas. A cada nova reprodução de uma música, um evento é gerado para 
atualizar o ranking. Como o volume de eventos é imenso, manter todas as músicas disponíveis 
ordenadas é inviável. Projete um algoritmo que processe um stream com 𝑛 eventos, contendo 
em cada um o identificador numérico da música, e produza o ranking de 𝑘 posições em 
𝑂(𝑛𝑙𝑜𝑔𝑘), onde 𝑘 ≪𝑛.


**Exemplos:**

| Eventos | Saída|
|---|---|
| `eventos = [5, 1, 2, 1, 3, 2, 1, 4, 2, 1]` e $k=3$ | `[(1, 4), (2, 3), (3, 1)]` |
| `eventos = [30, 10, 10, 20, 20, 10] ` e $k=2$ | `[(10, 3), (20, 2)]` |
| `eventos = [2, 5, 5, 5, 5, 1, 3]` e $k=4$ | `[(5, 4), (1, 1), (2, 1), (3, 1)]` |
| `eventos = [100, 200, 100, 300, 200, 100, 200]` e $k=1$ | `[(100, 3)]` |

````python
def min_heapfy_down_count(v:list, low:int, high:int):
    """
    Assumindo que v[low:high+1] é uma min heap válida com excessão da root, reposiciona v[low] para a posição correta
    Assume que as entradas da min heap são do tipo (id, key)

    Parameters
    ----------
    v : list
        Array com uma heap entre low e high, com entradas (id, key), com a root possivelmente incorreta
    low : int
        Início do intervalo da heap no array
    high : int
        Final do intervalo da heap no array
    """

    if (low >= high): return

    max_logico = high - low
    atual_logico = 0

    while (2*atual_logico + 1) <= max_logico:
        if (2*atual_logico + 2 > max_logico):
            if (v[2*atual_logico+1+low][1] < v[atual_logico+low][1]):
                v[2*atual_logico+1+low],v[atual_logico+low] = v[atual_logico+low], v[2*atual_logico+1+low] 
                atual_logico = 2*atual_logico+1
            else:
                break # está na posição correta
        else:
            if (v[2*atual_logico+1+low][1] < v[2*atual_logico+2+low][1]):
                max_child_pos = 2*atual_logico+1
                max_child_key = v[max_child_pos+low][1]
            else:
                max_child_pos = 2*atual_logico+2
                max_child_key = v[max_child_pos+low][1]

            if (v[atual_logico+low][1] > max_child_key):
                v[max_child_pos+low], v[atual_logico+low] = v[atual_logico+low], v[max_child_pos+low]
                atual_logico = max_child_pos
            else:
                break # está na posição correta

def min_heapfy_up_count(v:list, low, high):
    """
    Assumindo que v[low:high] é uma min heap válida, reposiciona v[high] para a posição correta
    Assume que as entradas da min heap são do tipo (id, key)

    Parameters
    ----------
    v : list
        Array com heap entre low e high, com entradas (id, key), com o último elemento fora de lugar
    low : int
        Início do intervalo da heap no array
    high : int
        Final do intervalo da heap no array
    """
    if (low>=high): return

    atual_logico = high - low
    pai_logico = (atual_logico-1) // 2
    while (pai_logico >= 0):
        if ((v[pai_logico+low][1] > v[atual_logico+low][1]) or
            (v[pai_logico+low][1] == v[atual_logico+low][1] and
             v[pai_logico+low][0] < v[atual_logico+low][0])):
            v[pai_logico+low], v[atual_logico+low] = v[atual_logico+low], v[pai_logico+low]
            atual_logico = pai_logico
            pai_logico = (atual_logico-1) // 2
        else:
            break # já está na posição correta

def heapfy_up_full_song(v:list, low:int, high:int):
    if (low>=high): return

    atual_logico = high - low
    pai_logico = (atual_logico-1) // 2
    while (pai_logico >= 0):
        if ((v[pai_logico+low][1] < v[atual_logico+low][1]) or 
            (v[pai_logico+low][1] == v[atual_logico+low][1] and
             v[pai_logico+low][0] > v[atual_logico+low][0])):
            v[pai_logico+low], v[atual_logico+low] = v[atual_logico+low], v[pai_logico+low]
            atual_logico = pai_logico
            pai_logico = (atual_logico-1) // 2
        else:
            break # já está na posição correta

def heapfy_down_full_song(v:list, low:int, high:int):
    if (low >= high): return

    max_logico = high - low
    atual_logico = 0

    while (2*atual_logico + 1) <= max_logico:
        if (2*atual_logico + 2 > max_logico):
            if ((v[2*atual_logico+1+low][1] > v[atual_logico+low][1]) or
                (v[2*atual_logico+1+low][1] == v[atual_logico+low][1] and
                 v[2*atual_logico+1+low][0] < v[atual_logico+low][0])):
                v[2*atual_logico+1+low],v[atual_logico+low] = v[atual_logico+low], v[2*atual_logico+1+low] 
                atual_logico = 2*atual_logico+1
            else:
                break # está na posição correta
        else:
            if ((v[2*atual_logico+1+low][1] > v[2*atual_logico+2+low][1]) or
                (v[2*atual_logico+1+low][1] == v[2*atual_logico+2+low][1] and
                 v[2*atual_logico+1+low][0] < v[2*atual_logico+2+low][0])):
                max_child_pos = 2*atual_logico+1
                max_child_id = v[max_child_pos+low][0]
                max_child_count = v[max_child_pos+low][1]
            else:
                max_child_pos = 2*atual_logico+2
                max_child_id = v[max_child_pos+low][0]
                max_child_count = v[max_child_pos+low][1]

            if ((v[atual_logico+low][1] < max_child_count) or
                (v[atual_logico+low][1] == max_child_count and
                 v[atual_logico+low][0] > max_child_id)):
                v[max_child_pos+low], v[atual_logico+low] = v[atual_logico+low], v[max_child_pos+low]
                atual_logico = max_child_pos
            else:
                break # está na posição correta

def questao_03(v:list, k:int) -> list:
    # iteração pelas n reproduções, montando dicionário de frequências e achando o máximo ~ O(n)
    max_count = 0
    song_count = dict()
    for song in v:
        if song in song_count:
            song_count[song] += 1
        else:
            song_count[song] = 1

        if (song_count[song] > max_count):
            max_count = song_count[song]

    songs_heap = list()
    if (k <= 0): return songs_heap

    atual = 0
    for song, count in song_count.items():
        # montar a min heap com as k primeiras músicas ~ O(log k) por música
        if (atual < k):
            songs_heap.append((song, count))
            min_heapfy_up_count(songs_heap, 0, atual)
            atual += 1
            continue

        # se a música tiver mais reproduções que o mínimo, adiciona e reorganiza a heap ~ O(log k) por música
        if ((count > songs_heap[0][1]) or
            (count == songs_heap[0][1] and
             song < songs_heap[0][0])):
            songs_heap[0] = (song, count)
            min_heapfy_down_count(songs_heap, 0, atual-1)

    ranks_heap = list()
    for i in range(len(songs_heap)):
        ranks_heap.append(songs_heap[i])
        heapfy_up_full_song(ranks_heap, 0, i)

    ranks = list()
    for i in range(len(ranks_heap)):
        ranks.append(ranks_heap[0])
        ranks_heap[0] = ranks_heap[k-i-1]
        heapfy_down_full_song(ranks_heap, 0, k-i-1)

    return ranks

print(questao_03([5, 1, 2, 1, 3, 2, 1, 4, 2, 1], 3))
print(questao_03([30, 10, 10, 20, 20, 10], 2))
print(questao_03([2, 5, 5, 5, 5, 1, 3], 4))
print(questao_03([100, 200, 100, 300, 200, 100, 200], 1))
````

### Questão 04
Considerando a tabela hash com endereçamento aberto, responda as perguntas a seguir:

**a)** Explique como é a sua estrutura de dados, e como funcionam os algoritmos de busca, 
inserção e remoção. 

**Resposta:** A tabela hash é uma estrutura de dados pensada para otimizar o armazenamento, busca e atualização de pares <chave, valor>. Na implementação da tabela hash, temos um vetor de tamanho $M$ e uma função (hash em $O(1)$) que associa uma chave a um inteiro pertencente a $\set{0, 1, ..., M-1}$, o que permite associar uma chave à sua posição no vetor. Essa ideia permite buscar, inserir e remover em $O(1)$ no caso de cada chave possuir uma posição diferente, entretanto, o universo das chaves costuma ser maior que $M$, de modo que não se possa garantir que cada chave será a única associada à uma dada posição no vetor, logo, deve-se pensar em como lidar com o conflito. Na abordagem de endereçamento aberto esse conflito é tratado da seguinte maneira: 
- Na inserção de nova chave: calculamos o hash da nova chave, verificamos se a posição associada está ocupada, se estiver livre, colocamos o par <chave, valor> nessa posição, caso o contrário, verificamos as próximas posições sequencialmente e colocamos o par <chave, valor> na primeira posição livre. (a ideia é que, ao se buscar a chave, exista um bloco que possa ser percorrido até encontrar o valor de interesse ou uma posição vazia)
- Na remoção de uma chave: calculamos o hash da chave, verificamos se a chave está na posição associada, se estiver, removemos o que estava lá e inserímos uma flag de removido, caso o contrário, verificamos as próximas posições sequencialmente e, quando encontrarmos a chave, removemos o que estava lá e inserímos uma flag de removido. Caso se chegue em uma posição sem <chave, valor> e sem a flag de que houve remoção, sabemos que o bloco onde a chave poderia estar acabou e ela não está presente.
- Na busca de uma chave: calculamos o hash da chave, verificamos se a chave está na posição associada, se estiver, ótimo, caso o contrário, verificamos as próximas posições sequencialmente até encontrarmos a chave, caso encontremos alguma posição vazia mas com uma flag de removido, sabemos que mais coisas podem ter sido adicionadas após essa posição e continuamos buscado, caso a posição esteja vazia e não tenha a flag de removido, sabemos que o bloco onde a chave poderia estar acabou e ela não está presente.

**b)** Compare as estratégias de hash duplo e sondagem quadrática. 

**Resposta:** Diferentemente da estratégia de busca sequencial para as próximas posições possíveis de uma chave citada no item (a), sondagem quadrática e hash duplo são estratégias que definem uma forma alternativa de procurar por posições alternativas para uma chave, buscando evitar a existência de blocos densos ocupados no vetor e garantindo um melhor espalhamento das chaves. Para isso, a sondagem quadrática define que a $k$-ésima posição possível de uma chave será dada por $f(chave, k) = (hash(chave) + c_1\cdot k + c_2\cdot k^2) \mod M$. Desse modo, a sondagem quadrática garante que srão feitos pulos entre as posições possíveis para uma dada chave, espalhando-a melhor, entretanto, caso duas chaves distintas tenham a mesma hash, ambas as chaves terão a mesma sequência de posições possíveis, de forma que conflito ainda sejam comuns. Para lidar com esse problema, a estratégia de hash duplo define duas funções de hash e calcula a $k$-ésima posição possível para uma chave como $f(chave, k) = (hash_1(chave) + k\cdot hash_2(chave))\mod M$. Desse modo, as sequências de posições consideradas para duas chaves distintas tendem a ser diferentes, reduzindo o número esperado de conflitos quando comparada à estratégia de sondagem quadrática.

**c)** Explique o que é o processo de resize e re-hashing, e avalie a sua complexidade. 

**Resposta:** Independentemente da estratégia de sondagem escolhida, conforme o número de chaves armazenadas aumenta e se aproxima ao tamanho $M$ do vetor, o custo de todas essas operações, que é proporcional ao número de conflitos, aumenta, pois a chance de ocorrer um conflito aumenta. Logo, para manter o custo das operações de inserção, busca e remoção em $O(1)$,  tabelas hash implementam um processo de resize e re-hash. Esse processo consistem em, quando o número de chaves se aproxima de $M$, alocar um novo vetor de tamanho $M'$ e calcular um novo hash para todas as chaves já salvas, de forma a associá-las a uma posição no novo vetor. Como é preciso calcular um novo hash para cada uma das $n$ chaves armazenadas, e cada um desses calculos é $O(1)$, o custo dessa operação é $O(n)$ no total.
