### Questão 01 (2 pontos)

Seja $T(n)$ a complexidade do pior caso do algoritmo executado por `func`, determine uma função $f(n)$ tal que $T(n)=\Theta(f(n))$. A solução deverá explicar como $f(n)$ foi encontrada.

```c
void s(int v[], int i, int j) {
    int z = v[i];
    v[i] = v[j];
    v[j] = z;
}

int h(int v[], int a, int b) {
    int k = v[b];
    int f = a;
    for (int i=a; i<b; i++) {
        if (v[i] <= k) {
            s(v, i, f);
            f++;
        }
    }
    s(v, f, b);
    return f;
}

void g(int v[], int a, int b) {
    while (a < b) {
        int j = h(v, a, b);
        if (j - a < b - j) {
            g(v, a, j - 1);
            a = j + 1;
        } else {
            g(v, j + 1, b);
            b = j - 1;
        }
    }
}

bool func(int v[], int n, int k) {
    g(v, 0, n - 1);
    int i = 0;
    int j = n - 1;
    while (i < j) {
        int s = v[i] + v[j];
        if (s < k) {
            i++;
        } else if (s > k) {
            j--;
        } else {
            return true;
        }
    }
    return false;
}
```
